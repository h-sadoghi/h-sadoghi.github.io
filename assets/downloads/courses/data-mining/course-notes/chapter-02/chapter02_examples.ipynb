## 3. Simpson reversal

A beats B within each stratum but B wins with the observed pooled weights. This says nothing causal by itself.

```python
within, overall = simpson()
print("Within strata", within)
print("Pooled", overall)
assert within["easy"]["A"] > within["easy"]["B"]
assert within["hard"]["A"] > within["hard"]["B"]
assert overall["A"] < overall["B"]
```

## 4. Missingness selection bias

The full population values are known **only because this is an invented exercise**.

```python
full, observed = missing_bias()
print("Complete-data mean", full)
print("Observed-data mean", observed)
print("Relative increase:", (observed-full)/full)
```
