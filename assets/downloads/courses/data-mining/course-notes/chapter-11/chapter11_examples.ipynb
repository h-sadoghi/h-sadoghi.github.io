## خواندن دادهٔ ساختگی و اصلاح چندآزمونی

```python
with (BASE/'pvalues_synthetic.csv').open(encoding='utf-8',newline='') as f:
    rows=list(csv.DictReader(f))
p=[float(r['p_value']) for r in rows]
assert len(p)==6
print('Original:',p)
```

## تفاوت Bonferroni و Holm و BH

```python
for name,fn in [('Bonferroni',bonferroni),('Holm',holm),('BH',benjamini_hochberg)]:
    q=fn(p)
    print(name, 'adjusted:',[round(x,5) for x in q], 'accepted indices:', discoveries(q,.05))
```

## FWER در پنجاه آزمون مستقل

```python
print('Independent null tests m=50: FWER=', 1-(1-.05)**50)
print('Expected number of type I errors=',50*.05)
```

## جایگشت دقیق شش مشاهده

```python
A=[8,9,10];B=[1,2,3]
print('Observed difference=',sum(A)/3-sum(B)/3)
print('Exact 2-sided permutation p=',exact_permutation_difference(A,B))
```

## مونت‌کارلو با قرارداد +1

```python
for rep in [999,4999]:
    diff,pval=permutation_difference(A,B,repetitions=rep,seed=22)
    print(rep, 'permutations =>',round(pval,5))
```

## بازهٔ صدکیِ Bootstrap

```python
print('Bootstrap percentile difference of means:',bootstrap_mean_difference(A,B,3000,seed=41))
```

## آزمون McNemar با رکوردهای زوجی

```python
with (BASE/'paired_predictions_synthetic.csv').open(encoding='utf-8',newline='') as f:
    pairs=list(csv.DictReader(f))
a=[int(t['model_A_correct']) for t in pairs]
b=[int(t['model_B_correct']) for t in pairs]
print('Discordant b,c and exact p:',mcnemar_exact(a,b))
```
