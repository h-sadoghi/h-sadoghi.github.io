# Chapter 2 — Digital Image Fundamentals
مثال‌های اجرایی مستقل برای نمونه‌برداری، کوانتیزه‌سازی و درون‌یابی.


```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
N=256
y,x=np.mgrid[-1:1:complex(N),-1:1:complex(N)]
f=np.exp(-4*(x*x+y*y)) + .55*np.exp(-45*((x-.35)**2+(y+.22)**2)) + .25*(np.sin(10*x+4*y)+1)
f=(f-f.min())/(f.max()-f.min())
plt.imshow(f,cmap='gray'); plt.axis('off');
```

## Sampling and quantization


```python
def quantize(img,bits):
    L=2**bits
    return np.round(img*(L-1))/(L-1)

fig,ax=plt.subplots(2,3,figsize=(9,6))
for j,step in enumerate([1,4,8]):
    for i,bits in enumerate([8,3]):
        g=quantize(f[::step,::step],bits)
        ax[i,j].imshow(g,cmap='gray',vmin=0,vmax=1,interpolation='nearest')
        ax[i,j].set_title(f'step={step}, bits={bits}')
        ax[i,j].axis('off')
plt.tight_layout()
```

## Interpolation
نکته: درون‌یابی برای نقاط جدید مقدار تخمینی می‌سازد؛ جزئیات از دست‌رفته را واقعاً بازسازی نمی‌کند.


```python
small=Image.fromarray(np.uint8(f[::8,::8]*255))
fig,ax=plt.subplots(1,3,figsize=(9,3))
for a,(name,m) in zip(ax,[('Nearest',Image.Resampling.NEAREST),('Bilinear',Image.Resampling.BILINEAR),('Bicubic',Image.Resampling.BICUBIC)]):
    a.imshow(small.resize((256,256),m),cmap='gray'); a.set_title(name); a.axis('off')
plt.tight_layout()
```
