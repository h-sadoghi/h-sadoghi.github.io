# فصل ۶: تفاوت ساده RGB، HSI و HSV
**نمونه کوتاه و قابل اجرا.**


```python
import colorsys
for name, rgb in [("Red", (255,0,0)), ("Yellow", (255,255,0)), ("Gray", (120,120,120))]:
    h, s, v = colorsys.rgb_to_hsv(*(c/255 for c in rgb))
    i = sum(rgb)/3
    print(f"{name}: RGB={rgb}, H={h*360:.0f} deg, HSV S={s:.2f}, I={i:.0f}, V={v*255:.0f}")
print("Note: H is undefined for Gray, even if the program prints 0.")
```

برای پیکسل قرمز، شدت میانگین `I=85` اما مقدار بیشینه `V=255` است. تفاوت دو مدل را در همین خروجی ببینید.
