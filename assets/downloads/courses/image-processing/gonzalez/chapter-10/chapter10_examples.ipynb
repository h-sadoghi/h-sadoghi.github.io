# فصل ۱۰ — مثال‌های اجرایی قطعه‌بندی تصویر
این Notebook همراه فصل است و مثال‌های مشتق Sobel، Otsu، رشد ناحیه و مقیاس‌بندی نمایش پس از تبدیل لگاریتمی را بازتولید می‌کند.

پیش‌نیازها: `numpy`, `matplotlib`, `Pillow`.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from chapter10_examples import synthetic_image, sobel_xy, otsu_threshold, region_grow, minmax_scale
```

## ۱) مشتق‌های Sobel در راستای x و y

```python
f = synthetic_image()
gx, gy, mag = sobel_xy(f)
fig, ax = plt.subplots(1, 4, figsize=(12, 3))
for a, im, title in zip(ax, [f, np.abs(gx), np.abs(gy), mag], ['input', '|Gx|', '|Gy|', 'magnitude']):
    a.imshow(im, cmap='gray'); a.set_title(title); a.axis('off')
plt.tight_layout(); plt.show()
```

## ۲) آستانه Otsu و واریانس بین‌کلاسی

```python
k, sigma_b = otsu_threshold(f)
print('Otsu threshold =', k)
plt.figure(figsize=(8, 3.5))
plt.plot(sigma_b); plt.axvline(k, linestyle='--')
plt.xlabel('threshold k'); plt.ylabel('between-class variance'); plt.grid(alpha=.25); plt.show()
```

## ۳) رشد ناحیه روی ماتریس آموزشی فصل

```python
toy = np.array([[11,10,12,80,81],[10,12,11,79,82],[9,11,12,78,80],[10,13,11,77,81],[12,10,9,76,83]], dtype=float)
mask = region_grow(toy, seed=(2,1), T=3)
print(mask.astype(int))
plt.figure(figsize=(4,4)); plt.imshow(mask, cmap='gray'); plt.title('grown region'); plt.show()
```

## ۴) چرا خروجی لگاریتم برای نمایش به نرمال‌سازی نیاز دارد؟
برای $z=\log(1+f)$ بازه عددی بسیار کوچک‌تر از بازه ۸ بیتی است. رابطه نمایش فصل، بازه را به $0\ldots255$ می‌برد.

```python
z = np.log1p(f)
print('raw log range:', float(z.min()), float(z.max()))
g = minmax_scale(z)
fig, ax = plt.subplots(1, 3, figsize=(10,3))
ax[0].imshow(f, cmap='gray', vmin=0, vmax=255); ax[0].set_title('original')
ax[1].imshow(z.astype(np.uint8), cmap='gray', vmin=0, vmax=255); ax[1].set_title('direct log -> uint8')
ax[2].imshow(g, cmap='gray', vmin=0, vmax=255); ax[2].set_title('min-max display scaling')
for a in ax: a.axis('off')
plt.tight_layout(); plt.show()
```
